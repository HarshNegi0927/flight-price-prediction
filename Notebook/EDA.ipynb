# Analysis of Outliers

```python
forest = IsolationForest(n_estimators=100, contamination=0.05, random_state=42)
```

```python
(
	train
	.assign(outlier=forest.fit_predict(
		train
		.drop(columns="price")
		.select_dtypes(include="number")
	))
	.query("outlier == -1")
	# .duration
	# .describe()
)
```

```python
(
    train
    .assign(
        iqr_outlier=lambda df_: df_["duration"].pipe(
            lambda s: ~s.between(
                s.quantile(0.25) - 1.5 * (s.quantile(0.75) - s.quantile(0.25)),
                s.quantile(0.75) + 1.5 * (s.quantile(0.75) - s.quantile(0.25))
            )
        )
    )
    .query("iqr_outlier") 
)
```

## Date of Journey

```python
train.date_of_journey
```

```python
eda_helper_functions.dt_univar_plots(
	data=train,
	var="date_of_journey",
	target="price"
)
```

```python
(
	train
	.groupby(pd.Grouper(key="date_of_journey", freq="M"))
	.price.mean()
	.to_frame()
	.set_axis(["March", "April", "May", "June"], axis=0)
	.rename_axis(index="month")
	.plot(
		kind="bar",
		figsize=(8, 4),
		edgecolor="black"
	)
)

ax = plt.gca()
eda_helper_functions.rotate_xlabels(ax)
```

```python
(
	train
	.groupby(pd.Grouper(key="date_of_journey", freq="W"))
	.price.mean()
	.to_frame()
	.rename_axis(index="week")
	.plot(
		kind="bar",
		figsize=(8, 4),
		edgecolor="black"
	)
)

ax = plt.gca()
eda_helper_functions.rotate_xlabels(ax)
```

```python
(
	train
	.groupby([pd.Grouper(key="date_of_journey", freq="M"), "source"])
	.price.mean()
	.unstack(fill_value=0)
	.set_axis(["March", "April", "May", "June"], axis=0)
	.rename_axis(index="month")
	.plot(
		kind="bar",
		figsize=(12, 4),
		edgecolor="black"
	)
)

ax = plt.gca()

ax.legend(
	loc="upper left",
	bbox_to_anchor=(1, 1)
)

eda_helper_functions.rotate_xlabels(ax)
```
